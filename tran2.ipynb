```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib
```

```python
df =pd.read_csv('medical_data.csv')
```

```python
plt.Figure()
sns.histplot(data=df ,x='age', kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
plt.Figure()
sns.histplot(data=df , x='bmi', kde=False)
plt.title('bmi Distribution')
plt.xlabel('bmi')
plt.ylabel('Count')
plt.show()
```

```python
plt.Figure()
sns.barplot(data=df , x='age')
plt.show()
```

```python
plt.figure()
sns.barplot(data=df , x='bmi')
plt.show()  
```

```python
smoker_counts=df['smoker'].value_counts().reset_index()
```

```python
plt.Figure()
plt.pie(
    smoker_counts['count'],
    labels=smoker_counts['smoker'],
    autopct='%1.1f%%',
)
plt.title('Smoker Distribution')
plt.show()
```

#/bivariate analysis
2 colum

```python
plt.Figure()
sns.scatterplot(data=df , x='age', y='bmi', hue='smoker')
plt.show()
```

```python
plt.Figure()
sns.boxplot(data=df , x='gender', y='bmi')
plt.title('BMI Distribution by Gender')
plt.show()
```
